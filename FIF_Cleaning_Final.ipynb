```python
import pandas as pd
```

## Reading dataset and check the info

```python
df = pd.read_csv("../../dataset/All/22232-FIF.csv")
```

```python
df
```

```python
df.shape
```

```python
df.info()
```

## Count unique values in each column

```python
for col in df.columns:
    n = len(pd.unique(df[col]))
    print("{:<20}: {:<}".format(col, n))
```

## Check if there is null values

```python
df.isnull().sum()
```

## Change time from epoch to datetime

### Convert the time to datetime

```python
df['timecreated'] = pd.to_datetime(df['timecreated'], unit='s')
df
```

### Check the range of the dataset

```python
# Find the minimum and maximum dates
min_date = df['timecreated'].min()
max_date = df['timecreated'].max()

print('Minimum date:', min_date)
print('Maximum date:', max_date)
```

## Remove uneccessary columns

```python
df = df.drop(columns = ['objecttable', 'objectid', 'contextid', 
                         'contextlevel', 'contextinstanceid', 'relateduserid',
                         'anonymous', 'other', 'origin', 'ip', 'realuserid'])
df
```

```python
df.isnull().sum()
```

```python
df.info()
```

## Check if there is duplicates data and remove it

### Check for duplicates

```python
duplicate_mask = df.duplicated(keep=False)

duplicate_rows = df[duplicate_mask]
duplicate_rows
```

### Drop duplicates

```python
df.drop_duplicates()
```

## Shortened the event name

```python
shortened = list(value.split('\\')[1] + "\\" +  value.split('\\')[-1] for value in df["eventname"])
df["event"] = shortened
df
```

# Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
```

### Distribution of Events Over Time

```python
events_by_day = df.groupby(df['timecreated'].dt.date).size()
```

```python
# PLot the distribution of events over time
plt.figure(figsize=(10, 6))
events_by_day.plot(kind='line', marker='o', color='b')
plt.title('Distribution of Events Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Events')
plt.grid(True)
plt.xticks(rotation=45) # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```

```python
# Filter data between two dates
filtered_df = df.loc[(df['timecreated'] < '2023-06-26')]
```

```python
filtered_events_by_day = filtered_df.groupby(filtered_df['timecreated'].dt.date).size()
```

```python
# PLot the distribution of events over time
plt.figure(figsize=(10, 6))
filtered_events_by_day.plot(kind='line', marker='o', color='b')
plt.title('Distribution of Events Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Events')
plt.grid(True)
plt.xticks(rotation=45) # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```

```python
# Group events by educational level and calculate the frequency of events for each level
events_by_edulevel = filtered_df.groupby('edulevel').size()
```

```python
# Plot the distribution of events across different educational levels
plt.figure(figsize=(10, 6))
events_by_edulevel.plot(kind='bar', color='salmon')
plt.title('Distribution of Events Across Educational Levels')
plt.xlabel('Educational Level')
plt.ylabel('Frequency of Events')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Admin EDA

```python
# Select the admin data
admin_df = filtered_df[filtered_df['edulevel'] == 0]
# admin_df
```

### Count how many unique values in each column

```python
for col in admin_df.columns:
    n = len(pd.unique(admin_df[col]))
    print("{:<20}: {:<}".format(col, n))
```

### Events over time

```python
admin_events_by_day = admin_df.groupby(admin_df['timecreated'].dt.date).size()
```

```python
# Calculate the average number of events per day
average_events_per_day = admin_events_by_day.mean()

# PLot the distribution of events over time
plt.figure(figsize=(10, 6))
admin_events_by_day.plot(kind='line', marker='o', color='b')
plt.axhline(y=average_events_per_day, color='r', linestyle='--', label='Average Events per Day') # Add the average line
plt.title('Distribution of Events Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Events')
plt.grid(True)
plt.xticks(rotation=45) # Rotate x-axis labels for better readability
plt.legend() # Show legend
plt.tight_layout()
plt.show()
```

### Top 5 and bottom 5 of event frequencies

```python
event_counts = admin_df['event'].value_counts()

admin_top5_events = event_counts.head(5)
admin_bottom5_events = event_counts.tail(5)
```

```python
admin_combined_events = pd.concat([admin_top5_events, admin_bottom5_events])
```

```python
plt.figure(figsize=(12, 6))
bars = admin_combined_events.plot(kind='bar', color=['skyblue']*5 + ['salmon']*5)
plt.title('Top 5 and Bottom 5 Events by Frequency (Admin)')
plt.xlabel('Events')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()

for i, v in enumerate(admin_combined_events):
    bars.text(i, v + 0.2, str(v), ha='center', va='bottom')
    
plt.show()
```

```python
plt.figure(figsize=(12, 6))
bars = admin_combined_events.plot(kind='bar', color=['skyblue']*5 + ['salmon']*5)
plt.title('Top 5 and Bottom 5 Events by Frequency (Admin)')
plt.xlabel('Events')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()

for i, v in enumerate(admin_combined_events):
    bars.text(i, v + 0.2, str(v), ha='center', va='bottom')
    
plt.show()
```

```python
# Get the top 5 events
event_counts = lecturer_df['event'].value_counts()
lecturer_top5_events = event_counts.head(5).index

# Filter the dataframe to include only the top 5 events
top5_df = lecturer_df[lecturer_df['event'].isin(lecturer_top5_events)]

# Convert 'timecreated' to datetime if it's not already
top5_df['timecreated'] = pd.to_datetime(top5_df['timecreated'])

# Extract the date from 'timecreated'
top5_df['date'] = top5_df['timecreated'].dt.date

# Group by 'event' and 'date' and count occurrences
event_by_day = top5_df.groupby(['date', 'event']).size().unstack(fill_value=0)

# Plot the data
plt.figure(figsize=(12, 8))
for event in lecturer_top5_events:
    plt.plot(event_by_day.index, event_by_day[event], label=event)

plt.title('Top 5 Events by Day')
plt.xlabel('Date')
plt.ylabel('Count')
plt.legend(title='Event')
plt.grid(True)
plt.show()
```
